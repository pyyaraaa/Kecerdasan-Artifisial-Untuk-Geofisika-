dan dosen pembimbing yang menjelaskan secara sistematis, akurat, dan mudah dipahami.

geolistrik VES 1D, mulai dari perhitungan faktor geometri   dan resistivitas semu, deskripsi kurva sounding, penyusunan 
model lapisan, interpretasi keberadaan akuifer dangkal,     
hingga penyusunan kesimpulan akhir.

Studi kasus ini bertujuan untuk menduga keberadaan dan 
kedalaman akuifer dangkal pada sebuah titik sounding   
sebagai tahap awal eksplorasi air tanah.               

Kondisilokasi:

 ### **Promnt 2 Metode Seismik Refleksi**
 ### **(*Identifikasi struktur bawah permukaan menggunakan seismik refleksi 2D di suatu daerah*)**

**ROLE**

Kamu adalah ahli geofisika dengan spesialisasi metode seismik refleksi
(akuisisi, pengolahan, dan interpretasi) sekaligus dosen pembimbing yang
menjelaskan secara sistematis, akurat, dan mudah dipahami.

**TASK**

Susun kajian metode seismik refleksi yang mencakup: (1) prinsip dasar,
(2) parameter akuisisi data, (3) alur pengolahan data, (4) analisis kecepatan,
(5) interpretasi seismik, (6) kelebihan dan keterbatasan metode, serta
(7) STUDI KASUS: analisis data seismik 2D pada bagian INPUT, langkah demi
langkah, sampai kesimpulan geologi.

CONTEXT
Kajian ini untuk [tugas kuliah / laporan penelitian] dengan tujuan eksplorasi
hidrokarbon tahap awal. Pembaca adalah mahasiswa S1 Geofisika yang memahami
dasar gelombang dan geologi, tetapi belum mendalami seismik refleksi. Studi
kasus dipakai untuk menunjukkan bagaimana teori diterapkan pada data.

**INPUT**

Judul: Identifikasi struktur bawah permukaan menggunakan seismik refleksi 2D
di [nama cekungan/daerah]
Kondisi geologi: [cekungan sedimen, perlapisan batupasir-serpih, diduga ada
struktur antiklin dan sesar]
Parameter akuisisi (FIKTIF):
- Survei darat 2D, sumber: dinamit/vibroseis
- Jarak antar geophone: 25 m; jarak antar shot: 50 m
- Jumlah kanal: 120 (split-spread)
- Sampling interval: 2 ms; panjang rekaman: 3 detik
Hasil analisis kecepatan pada CMP 300 (FIKTIF):
Horizon | TWT (ms) | V_rms (m/s)
A       | 400      | 1800
B       | 850      | 2100
C       | 1300     | 2400
Data tambahan (FIKTIF):
- Densitas lapisan di atas A = 2,0 g/cc; lapisan A-B = 2,2 g/cc
- Horizon B tampak melengkung (crest) pada CMP 300 dengan TWT 850 ms dan
  TWT 900 ms pada sayap
- Terlihat pergeseran (offset) horizon B sekitar 40 ms pada CMP 320,
  diduga sesar normal
Software: [Seismic Unix / Petrel / OpendTect / lainnya]


**CONSTRAINTS**

- Bahasa Indonesia baku; istilah asing dicetak miring dan diberi padanan.
- Panjang maksimal [2.000] kata.
- Jangan mengarang referensi atau nilai di luar data yang diberikan. Jika
  tidak yakin, tulis "perlu diverifikasi".
- Pada studi kasus, jangan menyimpulkan lebih dari yang didukung data; tandai
  bahwa data bersifat ilustratif.
- Rumus ditulis lengkap beserta arti simbol dan satuannya (ingat: TWT adalah
  waktu tempuh pergi-pulang).

**OUTPUT**

1. Pendahuluan singkat (1 paragraf)
2. Prinsip dasar (hukum Snell, koefisien refleksi, impedansi akustik,
   waktu tempuh pergi-pulang/TWT)
3. Parameter akuisisi (jarak geophone, shot, fold, sampling, resolusi)
4. Alur pengolahan data (tabel berurutan: tahap, tujuan, contoh masalah
   yang diatasi), mulai dari geometri, koreksi statik, dekonvolusi, analisis
   kecepatan, NMO, stacking, hingga migrasi
5. Tabel perbandingan seismik refleksi vs refraksi
6. Studi kasus, dengan urutan:
   a. Ringkasan kasus dan tujuan
   b. Perhitungan fold (tunjukkan rumus dan hasilnya)
   c. Konversi TWT ke kedalaman horizon A
   d. Kecepatan interval antar horizon dengan persamaan Dix, lalu
      ketebalan tiap lapisan (tabel: lapisan, kecepatan interval, ketebalan)
   e. Koefisien refleksi pada batas lapisan yang datanya tersedia
   f. Interpretasi struktur (antiklin dan sesar) beserta alternatif tafsirnya
   g. Rekomendasi tindak lanjut (mis. seismik 3D, pengeboran, data sumur)
7. Keterbatasan dan sumber ketidakpastian (multiple, noise, kesalahan
   kecepatan, resolusi vertikal, pull-up/pull-down)
8. Kesimpulan

**QUALITY CRITERIA**

- Akurat secara ilmiah dan konsisten dengan literatur standar geofisika.
- Studi kasus runtut dari data mentah ke kesimpulan, dan tiap kesimpulan
  menyebut data pendukungnya.
- Membedakan jelas domain waktu (ms) dan domain kedalaman (m).
- Menyebut bahwa struktur pada penampang waktu bisa terdistorsi oleh variasi
  kecepatan, sehingga perlu konversi kedalaman.
- Mudah diikuti mahasiswa yang ingin mengulang perhitungannya.

**OPTIONAL VERIFICATION**

Sebelum menjawab, periksa bahwa:
- semua perhitungan memakai TWT/2 untuk kedalaman dan satuan konsisten
  (ms ke detik),
- persamaan Dix dipakai dengan benar (V_rms, bukan V_int, sebagai masukan),
- tafsir antiklin dan sesar tidak melebihi dukungan data, dan semua batasan
  terpenuhi.
Tandai bagian yang perlu dicek dengan data sumur atau sumber primer.


 ### **Promnt 3 Metode GPR** 
 ### **(*Deteksi pipa bawah tanah dan rongga di bawah badan jalan menggunakan GPR di suatu daerah*)**

 **ROLE**

Kamu adalah ahli geofisika dengan spesialisasi Ground Penetrating Radar (GPR)
(akuisisi, pengolahan, dan interpretasi radargram) sekaligus dosen pembimbing
yang menjelaskan secara sistematis, akurat, dan mudah dipahami.

**TASK**

Susun kajian metode GPR yang mencakup: (1) prinsip dasar, (2) sifat
elektromagnetik material (permitivitas, konduktivitas), (3) parameter
akuisisi dan pemilihan frekuensi antena, (4) alur pengolahan data,
(5) interpretasi radargram, (6) kelebihan dan keterbatasan metode, serta
(7) STUDI KASUS: analisis data GPR pada bagian INPUT, langkah demi langkah,
sampai kesimpulan.

**CONTEXT**

Kajian ini untuk [tugas kuliah / laporan penelitian] dengan tujuan
investigasi dangkal untuk deteksi utilitas bawah tanah (pipa) dan rongga di
bawah jalan. Pembaca adalah mahasiswa S1 Geofisika/Teknik yang memahami dasar
gelombang elektromagnetik, tetapi belum mendalami GPR. Studi kasus dipakai
untuk menunjukkan bagaimana teori diterapkan pada data.

**INPUT**

Judul: Deteksi pipa bawah tanah dan rongga di bawah badan jalan menggunakan
GPR di [nama lokasi]
Kondisi lokasi: [jalan lingkungan, tanah urug pasir-lempung, muka air tanah
tidak diketahui]
Parameter akuisisi (FIKTIF):
- Antena terlindung (shielded) frekuensi tengah 250 MHz, mode reflection profiling
- Panjang profil 20 m; jarak antar trace 0,05 m
- Time window 60 ns; 512 sampel per trace
Data radargram (FIKTIF):
- Anomali hiperbola 1: apeks pada x = 6 m, TWT apeks = 20 ns; pada jarak
  horizontal 1,0 m dari apeks, TWT = 28,3 ns
- Anomali hiperbola 2: apeks pada x = 14 m, TWT apeks = 12 ns
  (kecepatan dianggap sama dengan anomali 1)
- Zona refleksi kuat dengan polaritas terbalik pada x = 10-12 m, TWT = 16 ns,
  diduga rongga berisi udara
- Reflektor planar menerus pada TWT = 40 ns sepanjang profil, diduga batas
  lapisan lempung basah (permitivitas relatif lapisan bawah = 25)
Software: [ReflexW / GPR-SLICE / RadExplorer / lainnya]

**CONSTRAINTS**

- Bahasa Indonesia baku; istilah asing dicetak miring dan diberi padanan.
- Panjang maksimal [2.000] kata.
- Jangan mengarang referensi atau nilai di luar data yang diberikan. Jika
  tidak yakin, tulis "perlu diverifikasi".
- Pada studi kasus, jangan menyimpulkan lebih dari yang didukung data; tandai
  bahwa data bersifat ilustratif.
- Rumus ditulis lengkap beserta arti simbol dan satuannya (ingat: TWT adalah
  waktu tempuh pergi-pulang; kecepatan dalam m/ns).

**OUTPUT**

1. Pendahuluan singkat (1 paragraf)
2. Prinsip dasar (persamaan Maxwell secara konseptual, permitivitas relatif,
   kecepatan gelombang, koefisien refleksi, atenuasi)
3. Tabel kisaran permitivitas relatif dan kecepatan berbagai material
   (udara, air, pasir kering, pasir basah, lempung, aspal/beton), dengan
   catatan bahwa nilainya bervariasi
4. Parameter akuisisi (frekuensi vs kedalaman penetrasi vs resolusi,
   jarak trace, time window)
5. Alur pengolahan data (tabel berurutan: tahap, tujuan, contoh masalah
   yang diatasi), mulai dari koreksi waktu-nol, dewow, gain, filter
   band-pass, background removal, hingga migrasi
6. Studi kasus, dengan urutan:
   a. Ringkasan kasus dan tujuan
   b. Perhitungan jumlah trace dan kedalaman maksimum teoritis
   c. Estimasi kecepatan dari hiperbola difraksi, lalu permitivitas relatif
   d. Konversi TWT ke kedalaman untuk semua anomali dan reflektor
   e. Perhitungan panjang gelombang dan resolusi vertikal (kriteria λ/4)
   f. Koefisien refleksi pada batas lapisan dan pada rongga udara,
      beserta arti tanda (polaritas)-nya
   g. Interpretasi tiap anomali beserta alternatif tafsirnya
   h. Rekomendasi tindak lanjut (mis. profil silang, frekuensi lain,
      penggalian uji, pengeboran)
7. Keterbatasan dan sumber ketidakpastian (atenuasi di lempung basah,
   clutter, kesalahan kecepatan, ambiguitas pipa vs batu vs akar, efek
   multiple)
8. Kesimpulan

**QUALITY CRITERIA**

- Akurat secara ilmiah dan konsisten dengan literatur standar GPR.
- Studi kasus runtut dari data mentah ke kesimpulan, dan tiap kesimpulan
  menyebut data pendukungnya.
- Membedakan jelas domain waktu (ns) dan domain kedalaman (m).
- Menyebut bahwa satu bentuk anomali (hiperbola) bisa berasal dari beberapa
  objek berbeda, sehingga perlu verifikasi langsung.
- Mudah diikuti mahasiswa yang ingin mengulang perhitungannya.

**OPTIONAL VERIFICATION**

Sebelum menjawab, periksa bahwa:
- kedalaman dihitung dengan TWT/2 dan satuan konsisten (ns dan m/ns),
- estimasi kecepatan dari hiperbola memakai persamaan yang benar,
- tanda koefisien refleksi sesuai arah kontras permitivitas,
- tafsir tidak melebihi dukungan data, dan semua batasan terpenuhi.
Tandai bagian yang perlu dicek dengan penggalian uji atau sumber primer.

 ### **Promnt 4 Metode Geomagnet**
 ### **(*Identifikasi anomali magnetik dugaan endapan bijih besi menggunakan metode geomagnet di suatu daerah*)**

**ROLE**

Kamu adalah ahli geofisika dengan spesialisasi metode geomagnet (magnetik)
(akuisisi, koreksi data, pengolahan, dan interpretasi anomali medan magnet)
sekaligus dosen pembimbing yang menjelaskan secara sistematis, akurat, dan
mudah dipahami.

**TASK**

Susun kajian metode geomagnet yang mencakup: (1) prinsip dasar,
(2) suseptibilitas dan magnetisasi batuan, (3) parameter akuisisi,
(4) koreksi data (variasi harian dan medan utama/IGRF), (5) pengolahan
lanjut (reduksi ke kutub, kelanjutan ke atas, sinyal analitik),
(6) interpretasi anomali, (7) kelebihan dan keterbatasan metode, serta
(8) STUDI KASUS: analisis data magnetik pada bagian INPUT, langkah demi
langkah, sampai kesimpulan geologi.

**CONTEXT**

Kajian ini untuk [tugas kuliah / laporan penelitian] dengan tujuan eksplorasi
mineral (dugaan endapan bijih besi bertipe magnetit). Pembaca adalah mahasiswa
S1 Geofisika/Geologi yang memahami dasar kemagnetan dan geologi, tetapi belum
mendalami metode ini. Studi kasus dipakai untuk menunjukkan bagaimana teori
diterapkan pada data.

**INPUT**

Judul: Identifikasi anomali magnetik dugaan endapan bijih besi menggunakan
metode geomagnet di [nama daerah]
Kondisi geologi: [batuan sedimen tertutup lapukan tipis, diduga ada intrusi
batuan beku di bawah permukaan]
Parameter akuisisi (FIKTIF):
- Alat: Proton Precession Magnetometer (PPM), pengukuran medan total
- Grid 200 m x 200 m; spasi lintasan 20 m; spasi titik 10 m
- Tinggi sensor 2 m; satu magnetometer base station, pembacaan tiap 10 menit
- Sensor dijauhkan dari benda logam, pagar besi, dan kendaraan
Data referensi (FIKTIF):
- Medan utama IGRF lokasi = 44.500 nT; inklinasi = -25°; deklinasi = 1°
- Nilai base station acuan (awal survei) = 44.620 nT
Data pengukuran (FIKTIF):
Stasiun | T_obs (nT) | Base station saat pengukuran (nT)
P-03    | 44.530     | 44.630   (titik latar/background)
P-07    | 44.980     | 44.640   (puncak anomali positif)
P-12    | 44.350     | 44.650   (anomali negatif, 60 m di sebelah selatan P-07)
Data tambahan (FIKTIF):
- Pada profil yang melewati P-07, jarak horizontal antara dua titik singgung
  lereng setengah-maksimum (half-slope) = 48 m
- Suseptibilitas magnetik tubuh anomali (dugaan dari pengukuran sampel)
  k = 0,5 (satuan SI)
Software: [Geosoft Oasis Montaj / Surfer / Magpick / lainnya]

**CONSTRAINTS**

- Bahasa Indonesia baku; istilah asing dicetak miring dan diberi padanan.
- Panjang maksimal [2.000] kata.
- Jangan mengarang referensi atau nilai di luar data yang diberikan. Jika
  tidak yakin, tulis "perlu diverifikasi".
- Pada studi kasus, jangan menyimpulkan lebih dari yang didukung data; tandai
  bahwa data bersifat ilustratif.
- Rumus ditulis lengkap beserta arti simbol dan satuannya (nT, A/m, SI);
  jelaskan hubungan 1 nT = 10^-9 T.

**OUTPUT**

1. Pendahuluan singkat (1 paragraf)
2. Prinsip dasar (medan magnet bumi, komponen F, H, Z, D, I, magnetisasi
   induksi dan remanen, suseptibilitas)
3. Tabel kisaran suseptibilitas berbagai material (magnetit, basalt,
   granit, batupasir, serpih, dll.), dengan catatan bahwa nilainya bervariasi
4. Parameter akuisisi (spasi lintasan/titik, orientasi lintasan, tinggi
   sensor, sumber noise)
5. Alur koreksi dan pengolahan data (tabel berurutan: tahap, tujuan, contoh
   masalah yang diatasi), mulai dari koreksi variasi harian, pengurangan
   IGRF, hingga reduksi ke kutub, kelanjutan ke atas, dan sinyal analitik
6. Studi kasus, dengan urutan:
   a. Ringkasan kasus dan tujuan
   b. Perhitungan jumlah lintasan, titik per lintasan, dan total titik
   c. Koreksi variasi harian dan anomali medan total (ΔT) untuk P-03, P-07,
      dan P-12 (tunjukkan rumus dan tiap langkah)
   d. Amplitudo puncak-ke-puncak anomali dipol (P-07 dan P-12) dan
      pemeriksaan posisi lobus negatif terhadap lobus positif untuk
      inklinasi negatif (belahan bumi selatan)
   e. Estimasi kedalaman dengan metode Peters (half-slope)
   f. Perhitungan kuat medan induksi H (A/m) dan magnetisasi induksi M = kH
   g. Interpretasi anomali beserta alternatif tafsirnya (magnetit, intrusi
      basa, magnetisasi remanen, objek logam modern)
   h. Catatan tentang kesesuaian reduksi ke kutub pada inklinasi rendah,
      dan alternatif (sinyal analitik atau reduksi ke ekuator)
   i. Rekomendasi tindak lanjut (mis. survei gravitasi, geolistrik/IP,
      pengeboran uji, pengukuran suseptibilitas sampel)
7. Keterbatasan dan sumber ketidakpastian (ambiguitas sumber, remanensi,
   noise geologi dekat permukaan, badai magnetik, keterbatasan metode Peters)
8. Kesimpulan

**QUALITY CRITERIA**

- Akurat secara ilmiah dan konsisten dengan literatur standar geofisika.
- Studi kasus runtut dari data mentah ke kesimpulan, dan tiap kesimpulan
  menyebut data pendukungnya.
- Membedakan jelas medan total (F), anomali (ΔT), dan medan utama (IGRF).
- Menyebut bahwa satu anomali magnetik dapat dihasilkan oleh banyak
  konfigurasi sumber yang berbeda (ambiguitas), sehingga perlu data
  pendukung.
- Mudah diikuti mahasiswa yang ingin mengulang perhitungannya.

**OPTIONAL VERIFICATION**

Sebelum menjawab, periksa bahwa:
- koreksi harian = (base saat pengukuran - base acuan), lalu dikurangkan,
  dan tanda hitungan konsisten,
- satuan konsisten (nT ke T, lalu ke A/m dengan μ0 = 4π x 10^-7 H/m),
- posisi lobus negatif sesuai dengan inklinasi yang diberikan,
