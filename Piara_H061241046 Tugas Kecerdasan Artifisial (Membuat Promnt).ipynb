### **Promnt 1 Metode Geolistrik** 
### **(*Pendugaan akuifer dangkal disuatu daerah menggunakan Vertical Electrical Sounding (VES) 1D, konfigurasi Schlumberger.*)**

Kamu adalah ahli geofisika dengan spesialisasi metode geolistrik (resistivitas)
dan dosen pembimbing yang menjelaskan secara sistematis, akurat, dan mudah dipahami.

**Task**                                                  

Selesaikan studi kasus geofisika berikut menggunakan metode 
geolistrik VES 1D, mulai dari perhitungan faktor geometri   dan resistivitas semu, deskripsi kurva sounding, penyusunan 
model lapisan, interpretasi keberadaan akuifer dangkal,     
hingga penyusunan kesimpulan akhir.     

**Context**

Studi kasus ini bertujuan untuk menduga keberadaan dan 
kedalaman akuifer dangkal pada sebuah titik sounding   
sebagai tahap awal eksplorasi air tanah.               

Kondisilokasi:                                                                                                                             
- Geologi: dataran aluvial dengan endapan pasir dan   
lempung                                                   
- Muka air tanah: belum diketahui  

**INPUT**
- Data pembanding: tidak tersedia data bor maupun sumuJudul: Pendugaan akuifer dangkal di daerah [nama daerah]

|---|---|

- Panjang maksimal [2.000] kata.
- Jangan mengarang referensi atau nilai di luar data yang diberikan. Jika tidak
  yakin, tulis "perlu diverifikasi".
- Pada studi kasus, jangan menyimpulkan lebih dari yang didukung data; tandai
  bahwa data bersifat ilustratif.
- Rumus ditulis lengkap beserta arti simbol dan satuannya.


**QUALITY CRITERIA**

- Akurat secara ilmiah dan konsisten dengan literatur standar geofisika.
- Studi kasus runtut dari data mentah ke kesimpulan, dan tiap kesimpulan
  menyebut data pendukungnya.
- Menyebut bahwa satu nilai resistivitas bisa mewakili beberapa litologi
  (mis. lempung basah vs. akuifer air payau).

**OPTIONAL VERIFICATION**

Sebelum menjawab, periksa bahwa:

6. Kesimpulan akhir yang menjawab tujuan studi kasus       

```python
import numpy as np
import matplotlib.pyplot as plt

# ===== INPUT (data fiktif) =====
AB2  = np.array([1.5, 3, 5, 10, 20, 40, 60, 100])   # AB/2 (m)
rho_a = np.array([45, 60, 85, 110, 70, 35, 28, 30]) # resistivitas semu (Ohm.m)
MN = 0.5  # jarak elektroda potensial (m), sesuai Input (MN <= AB/5 terpenuhi untuk AB/2 = 1,5 m)

# ===== 1. Faktor geometri Schlumberger =====
# K = pi * [ (AB/2)^2 - (MN/2)^2 ] / MN
K = np.pi * (AB2**2 - (MN/2)**2) / MN

# ===== 2. Hitung balik delta V / I =====
dV_I = rho_a / K   # (Ohm)

print("=== TABEL PERHITUNGAN VES (MN = %.1f m, sesuai Input) ===" % MN)
print(f"{'AB/2 (m)':>9} {'K (m)':>10} {'rho_a (Ohm.m)':>14} {'dV/I (Ohm)':>12}")
for a, k, r, v in zip(AB2, K, rho_a, dV_I):
    print(f"{a:9.1f} {k:10.2f} {r:14.1f} {v:12.4f}")

# ===== 3. Deskripsi kurva =====
i_max = np.argmax(rho_a)
i_min = np.argmin(rho_a)
print("\n=== DESKRIPSI KURVA ===")
print(f"Puncak  : rho_a = {rho_a[i_max]} Ohm.m pada AB/2 = {AB2[i_max]} m")
print(f"Minimum : rho_a = {rho_a[i_min]} Ohm.m pada AB/2 = {AB2[i_min]} m")
print(f"Nilai akhir: {rho_a[-1]} Ohm.m (selisih dari minimum = {rho_a[-1]-rho_a[i_min]} Ohm.m)")

# Deteksi tipe kurva sederhana (rho1 vs rho2 vs rho3)
r1, r2, r3 = rho_a[0], rho_a[i_max], rho_a[i_min]
if r1 < r2 > r3:
    tipe = "Tipe K (rho1 < rho2 > rho3)"
elif r1 > r2 < r3:
    tipe = "Tipe H (rho1 > rho2 < rho3)"
elif r1 < r2 < r3:
    tipe = "Tipe A (naik terus)"
else:
    tipe = "Tipe Q (turun terus)"
print("Tipe kurva :", tipe)

# ===== 4. Model lapisan awal (perkiraan kasar, belum inversi) =====
print("\n=== MODEL LAPISAN AWAL (perkiraan, perlu inversi IPI2Win) ===")
print("Lapisan 1: ~40-45 Ohm.m   -> tanah penutup / lempung pasiran lembap")
print("Lapisan 2: ~110-130 Ohm.m -> pasir/kerikil tak jenuh")
print("Lapisan 3: ~25-30 Ohm.m   -> pasir lempungan jenuh air ATAU lempung (ambigu)")

# ===== 5. Grafik kurva sounding =====
plt.figure(figsize=(7, 5))
plt.loglog(AB2, rho_a, 'o-', color='tab:blue', lw=2)
plt.xlabel("AB/2 (m)")
plt.ylabel("Resistivitas semu, rho_a (Ohm.m)")
plt.title("Kurva Sounding VES 1D - Schlumberger")
plt.grid(True, which="both", ls="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

 ### **Promnt 2 Metode Seismik Refleksi**
 ### **(*Identifikasi struktur bawah permukaan menggunakan seismik refleksi 2D di suatu daerah*)**

Kamu adalah ahli geofisika dengan spesialisasi metode seismik refleksi

Data tambahan (FIKTIF):
- Densitas lapisan di atas A = 2,0 g/cc; lapisan A-B = 2,2 g/cc

  TWT 900 ms pada sayap

  diduga sesar normal
Software: [Seismic Unix / Petrel / OpendTect / lainnya]


**CONSTRAINTS**

- Bahasa Indonesia baku; istilah asing dicetak miring dan diberi padanan.
- Panjang maksimal [2.000] kata.
- Jangan mengarang referensi atau nilai di luar data yang diberikan. Jika
  tidak yakin, tulis "perlu diverifikasi".
- Pada studi kasus, jangan menyimpulkan lebih dari yang didukung data; tandai
  bahwa data bersifat ilustratif.
- Rumus ditulis lengkap beserta arti simbol dan satuannya (ingat: TWT adalah
  waktu tempuh pergi-pulang).

**OUTPUT**

1. Pendahuluan singkat (1 paragraf)
2. Prinsip dasar (hukum Snell, koefisien refleksi, impedansi akustik,
   waktu tempuh pergi-pulang/TWT)
3. Parameter akuisisi (jarak geophone, shot, fold, sampling, resolusi)
4. Alur pengolahan data (tabel berurutan: tahap, tujuan, contoh masalah
   yang diatasi), mulai dari geometri, koreksi statik, dekonvolusi, analisis
   kecepatan, NMO, stacking, hingga migrasi
5. Tabel perbandingan seismik refleksi vs refraksi
6. Studi kasus, dengan urutan:
   a. Ringkasan kasus dan tujuan
   b. Perhitungan fold (tunjukkan rumus dan hasilnya)
   c. Konversi TWT ke kedalaman horizon A
   d. Kecepatan interval antar horizon dengan persamaan Dix, lalu
      ketebalan tiap lapisan (tabel: lapisan, kecepatan interval, ketebalan)
   e. Koefisien refleksi pada batas lapisan yang datanya tersedia
   f. Interpretasi struktur (antiklin dan sesar) beserta alternatif tafsirnya
   g. Rekomendasi tindak lanjut (mis. seismik 3D, pengeboran, data sumur)
7. Keterbatasan dan sumber ketidakpastian (multiple, noise, kesalahan
   kecepatan, resolusi vertikal, pull-up/pull-down)
8. Kesimpulan

**QUALITY CRITERIA**

- Akurat secara ilmiah dan konsisten dengan literatur standar geofisika.
- Studi kasus runtut dari data mentah ke kesimpulan, dan tiap kesimpulan
  menyebut data pendukungnya.
- Membedakan jelas domain waktu (ms) dan domain kedalaman (m).
- Menyebut bahwa struktur pada penampang waktu bisa terdistorsi oleh variasi
  kecepatan, sehingga perlu konversi kedalaman.
- Mudah diikuti mahasiswa yang ingin mengulang perhitungannya.

**OPTIONAL VERIFICATION**

Sebelum menjawab, periksa bahwa:
- semua perhitungan memakai TWT/2 untuk kedalaman dan satuan konsisten
  (ms ke detik),
- persamaan Dix dipakai dengan benar (V_rms, bukan V_int, sebagai masukan),
- tafsir antiklin dan sesar tidak melebihi dukungan data, dan semua batasan
  terpenuhi.
Tandai bagian yang perlu dicek dengan data sumur atau sumber primer.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import convolve

# =====================================================
# 1. PARAMETER AKUISISI (FIKTIF)
# =====================================================
dt = 0.002              # sampling interval 2 ms
T = 3.0                 # panjang rekaman 3 detik
nt = int(T / dt) + 1
t = np.arange(nt) * dt  # sumbu waktu (s)

dx_geophone = 25.0      # jarak antar geophone (m)
n_ch = 120              # jumlah kanal split-spread
dx_cmp = dx_geophone / 2

# =====================================================
# 2. DATA HASIL ANALISIS KECEPATAN PADA CMP 300
# =====================================================
horizon = ["A", "B", "C"]
twt_ms = np.array([400, 850, 1300])
vrms = np.array([1800, 2100, 2400])

# Konversi Dix: Vrms -> Vint
twt_s = twt_ms / 1000.0
vint = np.zeros(3)
vint[0] = vrms[0]
for i in range(1, 3):
    num = vrms[i]**2 * twt_s[i] - vrms[i-1]**2 * twt_s[i-1]
    den = twt_s[i] - twt_s[i-1]
    vint[i] = np.sqrt(num / den)

# Konversi waktu ke kedalaman (menggunakan Vint)
depth = np.zeros(3)
depth[0] = vint[0] * twt_s[0] / 2
for i in range(1, 3):
    depth[i] = depth[i-1] + vint[i] * (twt_s[i] - twt_s[i-1]) / 2

# Densitas, impedansi akustik, dan koefisien refleksi
rho_A = 2.0   # lapisan di atas A (g/cc)
rho_AB = 2.2  # lapisan A-B (g/cc)
print("=" * 60)
print("HASIL ANALISIS KECEPATAN (CMP 300)")
print("=" * 60)
print(f"{'Horizon':<8}{'TWT(ms)':<10}{'Vrms':<10}{'Vint(Dix)':<12}{'Depth(m)':<10}")
for i in range(3):
    print(f"{horizon[i]:<8}{twt_ms[i]:<10}{vrms[i]:<10}{vint[i]:<12.1f}{depth[i]:<10.1f}")

# Impedansi akustik sederhana (rho dalam g/cc dipakai relatif)
# Lapisan 1 (atas A): rho=2.0, V=Vrms_A ; Lapisan 2 (A-B): rho=2.2, V=Vint_B
# Lapisan 3 (B-C): rho diasumsikan 2.4
rho = np.array([2.0, 2.2, 2.4])
AI = rho * vint
RC = np.zeros(3)
RC[0] = (AI[0] - rho[0] * 1500) / (AI[0] + rho[0] * 1500)  # kontras dg lapisan lapuk
RC[1] = (AI[1] - AI[0]) / (AI[1] + AI[0])
RC[2] = (AI[2] - AI[1]) / (AI[2] + AI[1])
print("\nKoefisien refleksi (A, B, C):", np.round(RC, 4))

# =====================================================
# 3. MODEL STRUKTUR: ANTIKLIN + SESAR NORMAL
# =====================================================
n_cmp = 600
cmp_axis = np.arange(1, n_cmp + 1)
x = cmp_axis * dx_cmp

def horizon_time(base_ms, crest_amp_ms, cmp_center=300, width=120):
    """Antiklin: TWT lebih kecil di puncak (crest) dibanding sayap."""
    return base_ms - crest_amp_ms * np.exp(-((cmp_axis - cmp_center) ** 2) / (2 * width ** 2))

# Horizon B: crest 850 ms di CMP 300, sayap 900 ms
tA = horizon_time(400, 25)
tB = horizon_time(900, 50)      # 900 - 50 = 850 ms di crest
tC = horizon_time(1350, 50)

# Sesar normal di CMP 320: offset ~40 ms pada horizon B (dan merambat ke C)
fault_cmp = 320
offset_B = 40
offset_C = 35
tB = tB + np.where(cmp_axis >= fault_cmp, offset_B, 0)
tC = tC + np.where(cmp_axis >= fault_cmp, offset_C, 0)

# =====================================================
# 4. SEISMIK SINTETIK (KONVOLUSI RICKER WAVELET)
# =====================================================
def ricker(f, dt, length=0.128):
    tw = np.arange(-length / 2, length / 2, dt)
    a = (np.pi * f * tw) ** 2
    return (1 - 2 * a) * np.exp(-a)

wavelet = ricker(30, dt)  # frekuensi dominan 30 Hz

section = np.zeros((nt, n_cmp))
for j in range(n_cmp):
    refl = np.zeros(nt)
    for tt, r in zip([tA[j], tB[j], tC[j]], RC):
        idx = int(round((tt / 1000.0) / dt))
        if 0 <= idx < nt:
            refl[idx] += r
    section[:, j] = convolve(refl, wavelet, mode="same")

# Tambah noise acak
rng = np.random.default_rng(42)
section += 0.03 * np.max(np.abs(section)) * rng.standard_normal(section.shape)

# =====================================================
# 5. SHOT GATHER SINTETIK (Hiperbola NMO)
# =====================================================
offsets = np.linspace(-(n_ch / 2) * dx_geophone, (n_ch / 2) * dx_geophone, n_ch)
gather = np.zeros((nt, n_ch))
for k, off in enumerate(offsets):
    refl = np.zeros(nt)
    for t0, v, r in zip(twt_s, vrms, RC):
        tx = np.sqrt(t0**2 + (off / v) ** 2)
        idx = int(round(tx / dt))
        if idx < nt:
            refl[idx] += r
    gather[:, k] = convolve(refl, wavelet, mode="same")

# =====================================================
# 6. KOREKSI NMO PADA GATHER
# =====================================================
nmo = np.zeros_like(gather)
for k, off in enumerate(offsets):
    for i in range(nt):
        t0 = t[i]
        # interpolasi Vrms terhadap t0
        v = np.interp(t0, twt_s, vrms)
        tx = np.sqrt(t0**2 + (off / v) ** 2)
        idx = tx / dt
        i0 = int(np.floor(idx))
        if i0 + 1 < nt:
            w = idx - i0
            nmo[i, k] = (1 - w) * gather[i0, k] + w * gather[i0 + 1, k]

# =====================================================
# 7. VISUALISASI
# =====================================================
fig, axs = plt.subplots(2, 2, figsize=(16, 11))
fig.suptitle("Identifikasi Struktur Bawah Permukaan - Seismik Refleksi 2D (Data Fiktif)",
             fontsize=14, fontweight="bold")

clip = 0.6 * np.max(np.abs(section))

# (a) Shot gather
ax = axs[0, 0]
ax.imshow(gather, aspect="auto", cmap="gray",
          extent=[offsets[0], offsets[-1], T, 0],
          vmin=-0.6 * np.max(np.abs(gather)), vmax=0.6 * np.max(np.abs(gather)))
ax.set_title("(a) Shot Gather (Split-spread, 120 kanal)")
ax.set_xlabel("Offset (m)")
ax.set_ylabel("TWT (s)")

# (b) Setelah NMO
ax = axs[0, 1]
ax.imshow(nmo, aspect="auto", cmap="gray",
          extent=[offsets[0], offsets[-1], T, 0],
          vmin=-0.6 * np.max(np.abs(nmo)), vmax=0.6 * np.max(np.abs(nmo)))
ax.set_title("(b) Gather setelah Koreksi NMO")
ax.set_xlabel("Offset (m)")
ax.set_ylabel("TWT (s)")

# (c) Penampang seismik stack
ax = axs[1, 0]
ax.imshow(section, aspect="auto", cmap="seismic",
          extent=[cmp_axis[0], cmp_axis[-1], T, 0],
          vmin=-clip, vmax=clip)
ax.set_ylim(1.8, 0)
ax.axvline(300, color="lime", ls="--", lw=1, label="CMP 300")
ax.axvline(320, color="yellow", ls="--", lw=1, label="Sesar (CMP 320)")
ax.set_title("(c) Penampang Seismik Stack 2D")
ax.set_xlabel("CMP")
ax.set_ylabel("TWT (s)")
ax.legend(loc="lower right")

# (d) Interpretasi horizon
ax = axs[1, 1]
ax.plot(cmp_axis, tA, "b-", lw=2, label="Horizon A")
ax.plot(cmp_axis, tB, "g-", lw=2, label="Horizon B")
ax.plot(cmp_axis, tC, "r-", lw=2, label="Horizon C")
ax.axvline(fault_cmp, color="k", ls="--", label="Sesar normal")
ax.annotate("Crest antiklin\n(850 ms)", xy=(300, 850), xytext=(380, 700),
            arrowprops=dict(arrowstyle="->"), fontsize=10)
ax.annotate("Offset ~40 ms", xy=(fault_cmp, 880), xytext=(380, 1000),
            arrowprops=dict(arrowstyle="->"), fontsize=10)
ax.set_ylim(1700, 200)
ax.set_title("(d) Interpretasi Horizon (Antiklin & Sesar)")
ax.set_xlabel("CMP")
ax.set_ylabel("TWT (ms)")
ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import fftconvolve
from scipy.ndimage import uniform_filter1d

# =====================================================
# 1. PARAMETER AKUISISI (FIKTIF)
# =====================================================
f_c = 0.25                 # frekuensi tengah antena 250 MHz = 0.25 GHz
L = 20.0                   # panjang profil (m)
dx = 0.05                  # jarak antar trace (m)
T_win = 60.0               # time window (ns)
ns = 512                   # jumlah sampel per trace

x = np.arange(0, L + dx, dx)          # 401 trace
ntr = len(x)
t = np.linspace(0, T_win, ns)         # sumbu waktu (ns)
dt = t[1] - t[0]
c = 0.3                               # kecepatan cahaya (m/ns)

print("=" * 60)
print("PARAMETER AKUISISI")
print("=" * 60)
print(f"Jumlah trace     : {ntr}")
print(f"Sampling interval: {dt:.4f} ns")

# =====================================================
# 2. ANALISIS HIPERBOLA -> KECEPATAN, PERMITIVITAS, KEDALAMAN
# =====================================================
# Anomali 1: apeks x=6 m, t0=20 ns; pada jarak 1,0 m dari apeks t=28,3 ns
# t^2 = t0^2 + (2*dx_h/v)^2  ->  v = 2*dx_h / sqrt(t^2 - t0^2)
x1, t1_apex = 6.0, 20.0
dx_h, t_off = 1.0, 28.3
v = 2 * dx_h / np.sqrt(t_off**2 - t1_apex**2)     # m/ns
eps_atas = (c / v) ** 2

# Anomali 2 (kecepatan dianggap sama)
x2, t2_apex = 14.0, 12.0

# Kedalaman = v * t0 / 2
d1 = v * t1_apex / 2
d2 = v * t2_apex / 2

# Rongga (reversed polarity) pada x=10-12 m, t=16 ns
t_void = 16.0
d_void = v * t_void / 2

# Reflektor planar pada t=40 ns (lapisan lempung basah, er = 25)
t_plane = 40.0
d_plane = v * t_plane / 2
eps_bawah = 25.0
v_bawah = c / np.sqrt(eps_bawah)

# Koefisien refleksi (polaritas relatif)
n1 = np.sqrt(eps_atas)
R_plane = (n1 - np.sqrt(eps_bawah)) / (n1 + np.sqrt(eps_bawah))   # negatif (er naik)
R_void = (n1 - 1.0) / (n1 + 1.0)                                  # positif (udara, er=1)

print("\n" + "=" * 60)
print("HASIL ANALISIS HIPERBOLA")
print("=" * 60)
print(f"Kecepatan lapisan atas      : {v:.4f} m/ns")
print(f"Permitivitas relatif atas   : {eps_atas:.2f}")
print(f"Anomali 1 (x=6 m)           : kedalaman = {d1:.2f} m")
print(f"Anomali 2 (x=14 m)          : kedalaman = {d2:.2f} m")
print(f"Rongga (x=10-12 m, 16 ns)   : kedalaman = {d_void:.2f} m")
print(f"Reflektor planar (40 ns)    : kedalaman = {d_plane:.2f} m")
print(f"Kecepatan lapisan bawah     : {v_bawah:.4f} m/ns (er=25)")
print(f"Koef. refleksi batas lempung: {R_plane:+.3f} (polaritas negatif)")
print(f"Koef. refleksi rongga udara : {R_void:+.3f} (polaritas terbalik)")
print(f"Resolusi vertikal (lambda/4): {v / f_c / 4:.2f} m")

# =====================================================
# 3. MODEL RADARGRAM SINTETIK
# =====================================================
def ricker(f, dt, half_len=4.0):
    tw = np.arange(-half_len, half_len + dt, dt)
    a = (np.pi * f * tw) ** 2
    return (1 - 2 * a) * np.exp(-a)

wavelet = ricker(f_c, dt)
alpha = 0.02   # atenuasi (1/ns)

refl = np.zeros((ns, ntr))

def add_spike(col, t_ns, amp):
    """Tempatkan spike dengan interpolasi linear antar sampel."""
    if t_ns <= 0 or t_ns >= T_win - dt:
        return
    idx = t_ns / dt
    i0 = int(np.floor(idx))
    w = idx - i0
    a = amp * np.exp(-alpha * t_ns)
    refl[i0, col] += (1 - w) * a
    refl[i0 + 1, col] += w * a

# Gelombang langsung (direct wave) -> akan dihilangkan saat processing
for j in range(ntr):
    add_spike(j, 1.5, 1.5)

# Hiperbola pipa 1 (diasumsikan pipa logam, R negatif) & pipa 2 (pipa plastik, R positif)
pipa = [(x1, t1_apex, -0.6), (x2, t2_apex, +0.3)]
for (xp, t0, R) in pipa:
    for j in range(ntr):
        dxh = x[j] - xp
        if abs(dxh) <= 3.0:
            tt = np.sqrt(t0**2 + (2 * dxh / v) ** 2)
            add_spike(j, tt, R * (t0 / tt))     # geometric spreading

# Rongga udara x=10-12 m, t=16 ns (polaritas terbalik, tapering di tepi)
for j in range(ntr):
    if 10.0 <= x[j] <= 12.0:
        taper = np.sin(np.pi * (x[j] - 10.0) / 2.0) ** 0.5
        add_spike(j, t_void, R_void * taper)

# Reflektor planar 40 ns
for j in range(ntr):
    add_spike(j, t_plane, R_plane)

radar_raw = fftconvolve(refl, wavelet[:, None], mode="same", axes=0)

# Noise acak
rng = np.random.default_rng(1)
radar_raw += 0.02 * np.max(np.abs(radar_raw)) * rng.standard_normal(radar_raw.shape)

# =====================================================
# 4. PROCESSING: DEWOW + MUTE DIRECT WAVE + GAIN
# =====================================================
# (a) Dewow: hilangkan komponen frekuensi rendah
win = int(10.0 / dt)
dewow = radar_raw - uniform_filter1d(radar_raw, size=win, axis=0)

# (b) Hilangkan direct wave: kurangi trace rata-rata hanya pada t < 10 ns
#     (background removal penuh akan menghapus reflektor planar 40 ns)
bg = dewow.mean(axis=1)
taper_bg = np.where(t < 8, 1.0, np.where(t < 10, 0.5 * (1 + np.cos(np.pi * (t - 8) / 2)), 0.0))
proc = dewow - (bg * taper_bg)[:, None]

# (c) Gain eksponensial (kompensasi atenuasi)
gain = np.exp(0.02 * t)
proc = proc * gain[:, None]

# =====================================================
# 5. VISUALISASI
# =====================================================
def t2d(tt):
    return tt * v / 2

def d2t(dd):
    return dd * 2 / v

fig, axs = plt.subplots(2, 2, figsize=(16, 11))
fig.suptitle("Deteksi Pipa Bawah Tanah & Rongga di Bawah Badan Jalan - GPR 250 MHz (Data Fiktif)",
             fontsize=14, fontweight="bold")

def show(ax, data, title, clipfrac=0.5):
    clip = clipfrac * np.max(np.abs(data))
    ax.imshow(data, aspect="auto", cmap="gray",
              extent=[0, L, T_win, 0], vmin=-clip, vmax=clip)
    ax.set_title(title)
    ax.set_xlabel("Jarak horizontal (m)")
    ax.set_ylabel("TWT (ns)")
    sec = ax.secondary_yaxis("right", functions=(t2d, d2t))
    sec.set_ylabel("Kedalaman (m), v = %.3f m/ns" % v)

# (a) Raw
show(axs[0, 0], radar_raw, "(a) Radargram Mentah (Raw)", 0.4)

# (b) Setelah processing
show(axs[0, 1], proc, "(b) Setelah Dewow, Mute Direct Wave & Gain", 0.4)

# (c) Interpretasi
ax = axs[1, 0]
show(ax, proc, "(c) Interpretasi Anomali", 0.4)
xx = np.linspace(-3, 3, 200)
for (xp, t0, nama, warna) in [(x1, t1_apex, "Pipa 1", "yellow"), (x2, t2_apex, "Pipa 2", "lime")]:
    ax.plot(xp + xx, np.sqrt(t0**2 + (2 * xx / v) ** 2), "--", color=warna, lw=1.5, label=nama)
    ax.plot(xp, t0, "o", color=warna)
ax.add_patch(plt.Rectangle((10, t_void - 2.5), 2, 5, fill=False, ec="cyan", lw=2, label="Rongga"))
ax.axhline(t_plane, color="red", ls="--", lw=1.2, label="Batas lempung basah")
ax.text(6.2, 18, "Pipa 1\n(~1,0 m)", color="yellow", fontsize=9)
ax.text(14.2, 10, "Pipa 2\n(~0,6 m)", color="lime", fontsize=9)
ax.text(10.0, 11.5, "Rongga", color="cyan", fontsize=9)
ax.set_ylim(T_win, 0)
ax.legend(loc="lower left", fontsize=8)

# (d) A-scan
ax = axs[1, 1]
for xi, nama, warna in [(6.0, "x = 6 m (apeks pipa 1)", "tab:orange"),
                        (11.0, "x = 11 m (rongga)", "tab:cyan"),
                        (3.0, "x = 3 m (tanpa anomali)", "tab:gray")]:
    j = np.argmin(np.abs(x - xi))
    tr = proc[:, j] / np.max(np.abs(proc))
    ax.plot(tr, t, color=warna, label=nama)
ax.set_ylim(T_win, 0)
ax.set_xlabel("Amplitudo ternormalisasi")
ax.set_ylabel("TWT (ns)")
ax.set_title("(d) A-scan (Trace Tunggal)")
ax.axhline(t_plane, color="red", ls="--", lw=0.8)
ax.axhline(t_void, color="cyan", ls=":", lw=0.8)
ax.grid(alpha=0.3)
ax.legend(fontsize=8)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import TwoSlopeNorm
from matplotlib.patches import Ellipse

# =====================================================
# 1. PARAMETER AKUISISI & DATA REFERENSI (FIKTIF)
# =====================================================
F_igrf = 44500.0       # medan utama IGRF (nT)
inc = -25.0            # inklinasi (derajat)
dec = 1.0              # deklinasi (derajat)
base_ref = 44620.0     # base station acuan awal survei (nT)
h_sensor = 2.0         # tinggi sensor (m)
grid_size = 200.0      # grid 200 m x 200 m
line_sp = 20.0         # spasi lintasan (m)
stn_sp = 10.0          # spasi titik (m)
k_susc = 0.5           # suseptibilitas (SI)
d_half = 48.0          # jarak half-slope (m)

stn = ["P-03", "P-07", "P-12"]
ket = ["Background", "Puncak anomali (+)", "Anomali negatif"]
t_obs = np.array([44530.0, 44980.0, 44650.0 - 300.0])   # 44350
base_t = np.array([44630.0, 44640.0, 44650.0])

n_line = int(grid_size / line_sp) + 1
n_stn = int(grid_size / stn_sp) + 1
print("=" * 70)
print("PARAMETER SURVEI")
print("=" * 70)
print(f"Jumlah lintasan : {n_line} (spasi {line_sp:.0f} m)")
print(f"Titik per lintasan : {n_stn} (spasi {stn_sp:.0f} m)")
print(f"Total stasiun   : {n_line * n_stn}")

# Komponen medan utama
H_comp = F_igrf * np.cos(np.radians(inc))
Z_comp = F_igrf * np.sin(np.radians(inc))
print(f"Komponen horizontal H ~ {H_comp:,.0f} nT, vertikal Z ~ {Z_comp:,.0f} nT (negatif = ke atas)")

# =====================================================
# 2. KOREKSI VARIASI HARIAN & ANOMALI
# =====================================================
drift = base_t - base_ref            # variasi harian pada saat pengukuran
t_corr = t_obs - drift               # medan total terkoreksi
anom = t_corr - F_igrf               # anomali terhadap IGRF
bg = anom[0]                         # background dari P-03
resid = anom - bg                    # anomali residual

print("\n" + "=" * 70)
print("KOREKSI VARIASI HARIAN & ANOMALI MAGNETIK")
print("=" * 70)
print(f"{'Stasiun':<9}{'T_obs':>9}{'Base(t)':>9}{'Drift':>8}{'T_corr':>9}{'Anomali':>9}{'Residual':>10}")
for i in range(3):
    print(f"{stn[i]:<9}{t_obs[i]:>9.0f}{base_t[i]:>9.0f}{drift[i]:>+8.0f}"
          f"{t_corr[i]:>9.0f}{anom[i]:>+9.0f}{resid[i]:>+10.0f}")
print(f"\nBackground (P-03) terhadap IGRF : {bg:+.0f} nT")
print(f"Base station acuan vs IGRF      : {base_ref - F_igrf:+.0f} nT (perlu dicek lokasi base)")

# =====================================================
# 3. ESTIMASI KEDALAMAN (HALF-SLOPE / PETERS)
# =====================================================
z_peters = d_half / 1.6              # kedalaman dari level sensor
z_simple = d_half / 2.0              # estimasi alternatif (batas bawah)
z_surface = z_peters - h_sensor      # kedalaman dari permukaan tanah

print("\n" + "=" * 70)
print("ESTIMASI KEDALAMAN (HALF-SLOPE)")
print("=" * 70)
print(f"Jarak half-slope d                 : {d_half:.0f} m")
print(f"Kedalaman Peters (d/1,6)           : {z_peters:.1f} m dari sensor")
print(f"Kedalaman alternatif (d/2)         : {z_simple:.1f} m dari sensor")
print(f"Kedalaman puncak tubuh (permukaan) : {z_surface:.1f} m (sensor {h_sensor:.0f} m di atas tanah)")

# =====================================================
# 4. MAGNETISASI INDUKSI DARI SUSEPTIBILITAS
# =====================================================
mu0 = 4 * np.pi * 1e-7
H_field = (F_igrf * 1e-9) / mu0                  # kuat medan (A/m)
M_ind = k_susc * H_field                         # magnetisasi induksi
k_eff = k_susc / (1 + k_susc / 3)                # koreksi demagnetisasi (pendekatan bola)
M_eff = k_eff * H_field

print("\n" + "=" * 70)
print("MAGNETISASI INDUKSI")
print("=" * 70)
print(f"Kuat medan H = F/mu0               : {H_field:.2f} A/m")
print(f"Magnetisasi M = k.H (k = {k_susc})      : {M_ind:.2f} A/m")
print(f"k efektif (demagnetisasi, bola)    : {k_eff:.4f} SI")
print(f"Magnetisasi efektif                : {M_eff:.2f} A/m")

# =====================================================
# 5. PETA ANOMALI SINTETIK (DIKUNCI OLEH DATA P-03, P-07, P-12)
# =====================================================
# Pasangan anomali dipolar: lobe positif di P-07, lobe negatif 60 m ke selatan (P-12).
# Pada inklinasi negatif (belahan bumi selatan) lobe negatif memang berada di selatan.
sigma = d_half / 2.0
x07, y07 = 100.0, 130.0
x12, y12 = 100.0, 70.0
w = np.exp(-np.hypot(x07 - x12, y07 - y12) ** 2 / (2 * sigma ** 2))
A1 = (resid[1] - w * resid[2]) / (1 - w ** 2)
A2 = (resid[2] - w * resid[1]) / (1 - w ** 2)

def model_resid(X, Y):
    g1 = A1 * np.exp(-((X - x07) ** 2 + (Y - y07) ** 2) / (2 * sigma ** 2))
    g2 = A2 * np.exp(-((X - x12) ** 2 + (Y - y12) ** 2) / (2 * sigma ** 2))
    return g1 + g2

def model_total(X, Y):
    return bg + model_resid(X, Y)

Ef, Nf = np.meshgrid(np.linspace(0, 200, 201), np.linspace(0, 200, 201))
resid_map = model_resid(Ef, Nf)

Es = np.arange(0, grid_size + 1, line_sp)
Ns = np.arange(0, grid_size + 1, stn_sp)
ES, NS = np.meshgrid(Es, Ns)

posisi = {"P-03": (20, 180), "P-07": (x07, y07), "P-12": (x12, y12)}
print("\n" + "=" * 70)
print("CEK MODEL TERHADAP DATA STASIUN")
print("=" * 70)
for nama, (px, py) in posisi.items():
    print(f"{nama}: model = {model_total(px, py):+.1f} nT, data = {anom[stn.index(nama)]:+.1f} nT")
print(f"Maks residual peta : {resid_map.max():+.0f} nT | Min residual peta : {resid_map.min():+.0f} nT")

# =====================================================
# 6. VISUALISASI
# =====================================================
fig, axs = plt.subplots(2, 2, figsize=(15, 11))
fig.suptitle("Anomali Magnetik Dugaan Endapan Bijih Besi - Geomagnet PPM (Data Fiktif)",
             fontsize=14, fontweight="bold")

# (a) Peta residual
ax = axs[0, 0]
norm = TwoSlopeNorm(vmin=-250, vcenter=0, vmax=450)
im = ax.imshow(resid_map, origin="lower", extent=[0, 200, 0, 200], cmap="RdBu_r", norm=norm)
ax.contour(Ef, Nf, resid_map, levels=np.arange(-200, 500, 50), colors="k", linewidths=0.4)
ax.plot(ES.ravel(), NS.ravel(), "k.", ms=1.5)
ax.axvline(100, color="k", ls="--", lw=0.8)
for nama, (px, py) in posisi.items():
    ax.plot(px, py, "o", mfc="yellow", mec="k", ms=8)
    ax.annotate(nama, (px, py), xytext=(px + 6, py + 6), fontsize=9, fontweight="bold")
plt.colorbar(im, ax=ax, label="Residual (nT)")
ax.set_title("(a) Peta Anomali Magnetik Residual")
ax.set_xlabel("Timur (m)")
ax.set_ylabel("Utara (m)")

# (b) Profil utara-selatan melalui P-07 dan P-12
ax = axs[0, 1]
yy = np.linspace(0, 200, 401)
prof = model_total(np.full_like(yy, 100.0), yy)
ax.plot(yy, prof, "b-", lw=2, label="Anomali terhadap IGRF")
ns_pts = np.arange(0, 201, 10)
ax.plot(ns_pts, model_total(np.full(ns_pts.shape, 100.0), ns_pts.astype(float)),
        "ko", ms=3, label="Titik ukur (10 m)")
ax.plot(130, anom[1], "r^", ms=10, label="P-07")
ax.plot(70, anom[2], "gv", ms=10, label="P-12")
yh = np.array([130 - d_half / 2, 130 + d_half / 2])
vh = model_total(np.full(2, 100.0), yh)
ax.plot(yh, vh, "s", color="orange", ms=8, label="Titik half-slope")
ax.annotate("", xy=(yh[0], vh.mean()), xytext=(yh[1], vh.mean()),
            arrowprops=dict(arrowstyle="<->", color="orange"))
ax.text(130, vh.mean() + 15, f"d = {d_half:.0f} m", ha="center", color="darkorange", fontsize=10)
ax.axhline(bg, color="gray", ls="--", lw=1, label=f"Background ({bg:+.0f} nT)")
ax.axhline(0, color="k", lw=0.6)
ax.set_title("(b) Profil Utara-Selatan (E = 100 m)")
ax.set_xlabel("Utara (m)")
ax.set_ylabel("Anomali (nT)")
ax.grid(alpha=0.3)
ax.legend(fontsize=8)

# (c) Pengaruh koreksi variasi harian
ax = axs[1, 0]
xb = np.arange(3)
wb = 0.35
before = t_obs - F_igrf
b1 = ax.bar(xb - wb / 2, before, wb, color="tab:gray", label="Sebelum koreksi harian (T_obs - IGRF)")
b2 = ax.bar(xb + wb / 2, anom, wb, color="tab:blue", label="Setelah koreksi harian")
for bars in (b1, b2):
    for r in bars:
        h = r.get_height()
        ax.text(r.get_x() + r.get_width() / 2, h + (10 if h >= 0 else -30),
                f"{h:+.0f}", ha="center", fontsize=9)
ax.axhline(0, color="k", lw=0.8)
ax.set_xticks(xb)
ax.set_xticklabels([f"{s}\n{k}" for s, k in zip(stn, ket)])
ax.set_ylabel("Anomali (nT)")
ax.set_title("(c) Efek Koreksi Variasi Harian")
ax.set_ylim(-260, 560)
ax.grid(alpha=0.3, axis="y")
ax.legend(fontsize=8, loc="upper right")

# (d) Penampang interpretasi skematik
ax = axs[1, 1]
ax.axhspan(0, 5, color="sandybrown", alpha=0.6, label="Lapukan tipis (asumsi)")
ax.axhspan(5, 80, color="khaki", alpha=0.35, label="Batuan sedimen")
ax.axhline(0, color="k", lw=1)
tubuh = Ellipse((100, z_surface + 15), width=d_half, height=30, color="firebrick", alpha=0.75,
                label="Dugaan tubuh bijih besi (skematik)")
ax.add_patch(tubuh)
ax.text(100, z_surface + 15, f"Bijih besi\nk = {k_susc} SI", ha="center", va="center",
        color="white", fontsize=9, fontweight="bold")
ax.plot(ns_pts, np.full(ns_pts.shape, -h_sensor, dtype=float), "kv", ms=3)
ax.plot(130, -h_sensor, "r^", ms=9)
ax.plot(70, -h_sensor, "gv", ms=9)
ax.text(130, -5, "P-07", ha="center", fontsize=9)
ax.text(70, -5, "P-12", ha="center", fontsize=9)
ax.annotate("", xy=(145, 0), xytext=(145, z_surface), arrowprops=dict(arrowstyle="<->"))
ax.text(148, z_surface / 2, f"~{z_surface:.0f} m", fontsize=10)
ax.set_xlim(0, 200)
ax.set_ylim(80, -9)
ax.set_title("(d) Penampang Interpretasi (Skematik)")
ax.set_xlabel("Utara (m)")
ax.set_ylabel("Kedalaman (m)")
ax.legend(fontsize=8, loc="lower left")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

# =====================================================
# 1. PARAMETER AKUISISI & DATA (FIKTIF)
# =====================================================
L = 400.0                 # panjang profil (m)
dx = 20.0                 # spasi stasiun (m)
lat = -6.0                # 6 derajat LS
g_base_abs = 978105.00    # g absolut base station (mGal)
r_base_10 = 2000.00       # pembacaan base 10:00 (mGal)
r_base_14 = 2000.40       # pembacaan base 14:00 (mGal)
t_stn = 12.0              # waktu pembacaan P-09 (jam)
r_stn = 2013.49           # pembacaan P-09 (mGal)
h = 50.0                  # elevasi P-09 (m)
TC = 0.20                 # koreksi medan (mGal)
rho_B = 2.67              # densitas Bouguer (g/cc)
CBA_regional = 38.00      # anomali Bouguer lengkap P-01 (regional)
x_half = 60.0             # jarak ke setengah maksimum (m)
rho_host = 2.40           # densitas batuan sekitar (g/cc)
rho_body = 3.40           # densitas tubuh anomali (g/cc)

n_stn = int(L / dx) + 1
x = np.arange(n_stn) * dx
x_peak = 8 * dx           # P-09 = stasiun ke-9 -> x = 160 m
print("=" * 64)
print("PARAMETER SURVEI")
print("=" * 64)
print(f"Jumlah stasiun : {n_stn} (P-01 s/d P-{n_stn:02d}), spasi {dx:.0f} m")
print(f"P-09 berada pada x = {x_peak:.0f} m dari P-01")

# =====================================================
# 2. KOREKSI DRIFT ALAT
# =====================================================
drift_rate = (r_base_14 - r_base_10) / 4.0          # mGal/jam
drift_stn = drift_rate * (t_stn - 10.0)             # drift saat 12:00
r_corr = r_stn - drift_stn
g_obs = g_base_abs + (r_corr - r_base_10)           # g observasi absolut

print("\n" + "=" * 64)
print("KOREKSI DRIFT & GRAVITASI OBSERVASI (P-09)")
print("=" * 64)
print(f"Laju drift                 : {drift_rate:.3f} mGal/jam")
print(f"Drift pada 12:00           : {drift_stn:.2f} mGal")
print(f"Pembacaan terkoreksi drift : {r_corr:.2f} mGal")
print(f"g observasi                : {g_obs:,.2f} mGal")

# =====================================================
# 3. GRAVITASI TEORITIS (GRS80) & KOREKSI-KOREKSI
# =====================================================
phi = np.radians(abs(lat))
g_teo = 978032.7 * (1 + 0.0053024 * np.sin(phi) ** 2 - 0.0000058 * np.sin(2 * phi) ** 2)

FAC = 0.3086 * h                       # koreksi udara bebas
BC = 0.04193 * rho_B * h               # koreksi Bouguer
anom_raw = g_obs - g_teo
FAA = anom_raw + FAC                   # anomali udara bebas
SBA = FAA - BC                         # anomali Bouguer sederhana
CBA = SBA + TC                         # anomali Bouguer lengkap

print("\n" + "=" * 64)
print("ANOMALI BOUGUER LENGKAP (P-09)")
print("=" * 64)
print(f"g teoritis (GRS80)         : {g_teo:,.2f} mGal")
print(f"g_obs - g_teo              : {anom_raw:+.2f} mGal")
print(f"Koreksi udara bebas (FAC)  : {FAC:+.2f} mGal")
print(f"Anomali udara bebas (FAA)  : {FAA:+.2f} mGal")
print(f"Koreksi Bouguer (BC)       : {BC:-.2f} mGal")
print(f"Anomali Bouguer sederhana  : {SBA:+.2f} mGal")
print(f"Koreksi medan (TC)         : {TC:+.2f} mGal")
print(f"Anomali Bouguer lengkap    : {CBA:+.2f} mGal")

# =====================================================
# 4. SEPARASI REGIONAL-RESIDUAL
# =====================================================
g_res_max = CBA - CBA_regional
print("\n" + "=" * 64)
print("ANOMALI RESIDUAL")
print("=" * 64)
print(f"Regional (P-01)            : {CBA_regional:.2f} mGal")
print(f"Residual maksimum (P-09)   : {g_res_max:+.2f} mGal")

# =====================================================
# 5. PEMODELAN BOLA
# =====================================================
# Kedalaman pusat bola: z = 1,305 * x_1/2
z = x_half / np.sqrt(2 ** (2 / 3) - 1)               # = 1,305 * x_half
d_rho = rho_body - rho_host                          # kontras densitas (g/cc)
# g_max = (4/3) pi G d_rho R^3 / z^2 -> konstanta 0,02796 (mGal, g/cc, m)
K = (4.0 / 3.0) * np.pi * 6.674e-11 * 1000 * 1e5
R = (g_res_max * z ** 2 / (K * d_rho)) ** (1 / 3)
top = z - R
vol = 4.0 / 3.0 * np.pi * R ** 3
M_excess = vol * d_rho * 1000.0                      # kg
M_total = vol * rho_body * 1000.0                    # kg

print("\n" + "=" * 64)
print("PEMODELAN BOLA")
print("=" * 64)
print(f"Kedalaman pusat bola (z)   : {z:.1f} m")
print(f"Kontras densitas           : {d_rho:.2f} g/cc")
print(f"Radius bola (R)            : {R:.1f} m")
print(f"Kedalaman puncak bola      : {top:.1f} m")
print(f"Volume bola                : {vol:,.0f} m3")
print(f"Massa excess               : {M_excess/1e3:,.0f} ton ({M_excess/1e9:.2f} juta ton)")
print(f"Massa total tubuh          : {M_total/1e3:,.0f} ton")

# =====================================================
# 6. PROFIL SINTETIK
# =====================================================
xf = np.linspace(0, L, 801)
def sphere_g(xx):
    return g_res_max * z ** 3 / ((xx - x_peak) ** 2 + z ** 2) ** 1.5

rng = np.random.default_rng(7)
res_stn = sphere_g(x) + 0.03 * rng.standard_normal(n_stn)   # noise 0,03 mGal
res_stn[8] = g_res_max                                      # P-09 dikunci ke hasil hitung
cba_stn = CBA_regional + res_stn
cba_stn[0] = CBA_regional                                   # P-01 = regional

print("\n" + "=" * 64)
print("CEK MODEL")
print("=" * 64)
print(f"Residual model di x = +/-{x_half:.0f} m dari puncak : {sphere_g(x_peak + x_half):.2f} mGal "
      f"(setengah maks = {g_res_max/2:.2f})")

# =====================================================
# 7. VISUALISASI
# =====================================================
fig, axs = plt.subplots(2, 2, figsize=(15, 10))
fig.suptitle("Anomali Gravitasi Dugaan Tubuh Bijih - Metode Gravitasi (Data Fiktif)",
             fontsize=14, fontweight="bold")

# (a) Tahapan koreksi pada P-09
ax = axs[0, 0]
labels = ["g_obs - g_teo", "+ Udara bebas", "- Bouguer", "+ Medan\n(CBA)"]
vals = [anom_raw, FAA, SBA, CBA]
cols = ["tab:gray", "tab:blue", "tab:orange", "tab:green"]
bars = ax.bar(labels, vals, color=cols)
for b, v in zip(bars, vals):
    ax.text(b.get_x() + b.get_width() / 2, v + 0.8, f"{v:.2f}", ha="center", fontsize=10)
ax.axhline(CBA_regional, color="red", ls="--", lw=1.2, label=f"Regional P-01 = {CBA_regional:.0f} mGal")
ax.set_ylabel("Anomali (mGal)")
ax.set_title("(a) Tahapan Koreksi di Stasiun P-09")
ax.set_ylim(0, 52)
ax.grid(alpha=0.3, axis="y")
ax.legend(fontsize=9)

# (b) Profil CBA + regional
ax = axs[0, 1]
ax.plot(xf, CBA_regional + sphere_g(xf), "b-", lw=2, label="Model CBA")
ax.plot(x, cba_stn, "ko", ms=5, label="Data stasiun")
ax.axhline(CBA_regional, color="red", ls="--", label="Regional")
ax.plot(x_peak, CBA, "r^", ms=11, label="P-09 (puncak)")
ax.set_xlabel("Jarak dari P-01 (m)")
ax.set_ylabel("CBA (mGal)")
ax.set_title("(b) Profil Anomali Bouguer Lengkap")
ax.grid(alpha=0.3)
ax.legend(fontsize=9)

# (c) Residual dan half-width
ax = axs[1, 0]
ax.plot(xf, sphere_g(xf), "b-", lw=2, label="Model bola")
ax.plot(x, res_stn, "ko", ms=5, label="Residual (data)")
ax.axhline(g_res_max / 2, color="gray", ls=":", lw=1)
for xx in (x_peak - x_half, x_peak + x_half):
    ax.plot(xx, g_res_max / 2, "s", color="orange", ms=8)
ax.annotate("", xy=(x_peak - x_half, g_res_max / 2), xytext=(x_peak, g_res_max / 2),
            arrowprops=dict(arrowstyle="<->", color="darkorange"))
ax.text(x_peak - x_half / 2, g_res_max / 2 + 0.04, f"x1/2 = {x_half:.0f} m",
        ha="center", color="darkorange", fontsize=10)
ax.set_xlabel("Jarak dari P-01 (m)")
ax.set_ylabel("Residual (mGal)")
ax.set_title("(c) Anomali Residual & Titik Setengah Maksimum")
ax.grid(alpha=0.3)
ax.legend(fontsize=9)

# (d) Penampang interpretasi
ax = axs[1, 1]
ax.axhspan(0, 5, color="sandybrown", alpha=0.6, label="Tanah penutup tipis (asumsi)")
ax.axhspan(5, 200, color="khaki", alpha=0.35, label=f"Batuan sedimen ({rho_host} g/cc)")
ax.axhline(0, color="k", lw=1)
ax.add_patch(Circle((x_peak, z), R, color="firebrick", alpha=0.75,
                    label=f"Tubuh bijih ({rho_body} g/cc)"))
ax.text(x_peak, z, f"R = {R:.0f} m", ha="center", va="center",
        color="white", fontsize=10, fontweight="bold")
ax.plot(x, np.full(n_stn, -3.0), "kv", ms=4)
ax.plot(x_peak, -3.0, "r^", ms=9)
ax.text(x_peak, -9, "P-09", ha="center", fontsize=9)
ax.annotate("", xy=(x_peak + 100, 0), xytext=(x_peak + 100, top), arrowprops=dict(arrowstyle="<->"))
ax.text(x_peak + 105, top / 2, f"~{top:.0f} m", fontsize=10)
ax.annotate("", xy=(x_peak - 100, 0), xytext=(x_peak - 100, z), arrowprops=dict(arrowstyle="<->"))
ax.text(x_peak - 150, z / 2, f"z = {z:.0f} m", fontsize=10)
ax.set_xlim(0, L)
ax.set_ylim(180, -15)
ax.set_aspect("equal")
ax.set_xlabel("Jarak dari P-01 (m)")
ax.set_ylabel("Kedalaman (m)")
ax.set_title("(d) Penampang Interpretasi (Model Bola)")
ax.legend(fontsize=8, loc="lower left")

plt.tight_layout()
plt.show()
```
